# Math Notes

**By [Surya Raj](https://suryaraj.com)**

Welcome to **Math Notes** — a growing, open collection of mathematics notes, derivations, and Python-backed examples.
Whether you are a student refreshing fundamentals, an engineer looking for a quick reference, or someone who simply loves mathematics, you are in the right place.

Topics covered include Arithmetic & Algebra, Calculus (ODEs, BVPs, PDEs), Linear Algebra, Geometry, and Number Theory — all with worked examples and interactive Python notebooks.

All problems use Python with the [NumPy](https://numpy.org) {cite:p}`vanderWalt2011` and [SciPy](https://www.scipy.org) ecosystem of packages {cite:p}`2020SciPy-NMeth`.
This interactive website is powered by [Jupyter Book](https://jupyterbook.org/).
